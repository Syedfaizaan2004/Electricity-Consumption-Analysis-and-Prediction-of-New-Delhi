# Import Libraries

```python
import pandas as pd
import numpy as np
import os
import glob
```

# Set Data Path

```python
data_path = "../Data/"
```

# View All Files

```python
all_files = glob.glob(os.path.join(data_path, "*", "*.csv"))
print("Total CSV Files:", len(all_files))
```

# Read and Merge All CSV Files

```python
df_list = []
for file in all_files:
    # Read CSV file
    temp_df = pd.read_csv(file)
    # Extract filename
    filename = os.path.basename(file)
    
    # Example filename:
    # delhi_01_04_2022.csv
    
    date_part = filename.replace("delhi_", "").replace(".csv", "")
    
    # Convert to datetime
    date = pd.to_datetime(date_part, format="%d_%m_%Y")
    
    # Add new Date column
    temp_df['Date'] = date
    
    # Append dataframe
    df_list.append(temp_df)

# Merge all dataframes
df = pd.concat(df_list, ignore_index=True)

print("All files merged successfully!")
```

```python
# Check shape
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Remove duplicates
duplicate_rows = df.duplicated().sum()
print("Duplicate Rows:", duplicate_rows)
df = df.drop_duplicates()
print("Duplicates Removed!")
```

# Create Time Features

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Day_Name'] = df['Date'].dt.day_name()
```

```python
df['Total_Consumption'] = (
    df['BRPL'] +
    df['BYPL'] +
    df['NDPL'] +
    df['NDMC'] +
    df['MES']
)

print("Total_Consumption column created successfully!")
```

```python
# Convert TIMESLOT to String
df['TIMESLOT'] = df['TIMESLOT'].astype(str)
```

# Save Cleaned Dataset

```python
output_path = "../Output/cleaned_data.csv"
df.to_csv(output_path, index=False)
print("Cleaned data saved successfully!")
```
